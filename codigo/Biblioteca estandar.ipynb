# 1. `datetime` — Fechas y horas

El módulo `datetime` permite trabajar con fechas, horas e intervalos de tiempo de forma sencilla.

````python
from datetime import datetime, date, timedelta

# Fecha y hora actual
ahora = datetime.now()
print(ahora)                          # 2024-01-15 10:30:45.123456
print(ahora.year, ahora.month, ahora.day)
print(ahora.strftime("%d/%m/%Y %H:%M"))  # Formato personalizado
````

````python
# Operaciones con fechas
hoy = date.today()
cumple = date(1995, 6, 20)
edad = (hoy - cumple).days // 365
print(f"Días desde el cumpleaños: {(hoy - cumple).days}")
print(f"Edad aproximada: {edad} años")

# Sumar y restar tiempo
mañana = hoy + timedelta(days=1)
hace_una_semana = hoy - timedelta(weeks=1)
print(f"Mañana: {mañana}")
print(f"Hace una semana: {hace_una_semana}")
````

# 2. `math` — Funciones matemáticas

Proporciona funciones matemáticas clásicas: trigonometría, logaritmos, raíces, constantes...

````python
import math

print(math.pi)           # 3.141592653589793
print(math.e)            # 2.718281828459045
print(math.sqrt(144))    # 12.0
print(math.factorial(10))  # 3628800
print(math.ceil(4.2))    # 5   (redondea hacia arriba)
print(math.floor(4.9))   # 4   (redondea hacia abajo)
print(math.log(100, 10)) # 2.0  (logaritmo en base 10)
print(math.log2(1024))   # 10.0 (logaritmo en base 2)
print(math.pow(2, 10))   # 1024.0
````

````python
# Trigonometría (ángulos en radianes)
print(math.sin(math.pi / 2))   # 1.0
print(math.cos(0))              # 1.0
print(math.degrees(math.pi))   # 180.0
print(math.radians(180))       # 3.14159...

# Comprobación de infinito y NaN
print(math.isinf(float('inf')))   # True
print(math.isnan(float('nan')))   # True
````

# 4. `collections` — Colecciones especializadas

Ofrece tipos de datos de colección más potentes que las listas y diccionarios estándar.

````python
from collections import Counter

# Counter: cuenta ocurrencias
palabras = ['sol', 'luna', 'sol', 'estrella', 'sol', 'luna']
contador = Counter(palabras)
print(contador)                    # Counter({'sol': 3, 'luna': 2, 'estrella': 1})
print(contador['sol'])             # 3
print(contador.most_common(2))     # [('sol', 3), ('luna', 2)]

# También funciona con strings
letras = Counter('mississippi')
print(letras)
````

````python
from collections import defaultdict

# defaultdict: diccionario con valor por defecto automático
grupos = defaultdict(list)
datos = [('frutas', 'manzana'), ('verduras', 'zanahoria'),
         ('frutas', 'pera'), ('verduras', 'brocoli')]
for categoria, item in datos:
    grupos[categoria].append(item)   # No lanza KeyError si la clave no existe
print(dict(grupos))
````

````python
from collections import namedtuple, deque

# namedtuple: tupla con nombres de campo (inmutable)
Punto = namedtuple('Punto', ['x', 'y'])
p = Punto(3, 4)
print(p.x, p.y)         # Acceso por nombre
print(p[0], p[1])       # También por índice

Persona = namedtuple('Persona', ['nombre', 'edad', 'ciudad'])
alumno = Persona('Ana', 22, 'Madrid')
print(alumno)

# deque: cola de doble extremo, eficiente en ambos extremos
cola = deque([1, 2, 3])
cola.appendleft(0)   # [0, 1, 2, 3]
cola.append(4)        # [0, 1, 2, 3, 4]
cola.popleft()        # Quita el 0
print(cola)
````

# 5. `time` — Medición de tiempo

Proporciona funciones relacionadas con el tiempo del sistema, útiles para medir rendimiento o introducir pausas.

````python
import time

# Tiempo actual en segundos desde epoch (1 enero 1970)
print(time.time())

# Medir tiempo de ejecución con perf_counter (más preciso)
inicio = time.perf_counter()
total = sum(range(1_000_000))
fin = time.perf_counter()
print(f"Suma completada en {fin - inicio:.4f} segundos")
print(f"Resultado: {total}")
````

````python
# time.sleep: pausa la ejecución
print('Inicio')
time.sleep(0.5)   # Pausa 0.5 segundos
print('Fin (0.5 segundos después)')

# Tiempo local desglosado
t = time.localtime()
print(f"Año: {t.tm_year}, Mes: {t.tm_mon}, Día: {t.tm_mday}")
print(time.strftime('%A, %d de %B de %Y', t))
````

# 6. `string` — Constantes y utilidades de texto

Contiene constantes útiles con conjuntos de caracteres y herramientas para trabajar con texto.

````python
import string

print(string.ascii_lowercase)   # 'abcdefghijklmnopqrstuvwxyz'
print(string.ascii_uppercase)   # 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
print(string.digits)            # '0123456789'
print(string.punctuation)       # signos de puntuación

# string.capwords: capitaliza cada palabra
frase = '  hola   mundo  python  '
print(string.capwords(frase))   # 'Hola Mundo Python'
````

````python
# Generar contraseñas aleatorias
import random

def generar_contrasena(longitud: int = 12) -> str:
    caracteres = string.ascii_letters + string.digits + string.punctuation
    return ''.join(random.choices(caracteres, k=longitud))

for _ in range(3):
    print(generar_contrasena())
````

````python
# Template strings: sustitución de variables segura
from string import Template

plantilla = Template('Hola $nombre, tienes $edad años y vives en $ciudad.')
resultado = plantilla.substitute(nombre='Ana', edad=25, ciudad='Sevilla')
print(resultado)

# safe_substitute: no lanza error si falta una variable
resultado2 = plantilla.safe_substitute(nombre='Luis')
print(resultado2)   # $edad y $ciudad quedan sin sustituir
````

# 7. `re` — Expresiones regulares

Permite buscar, validar y transformar texto usando patrones. Es muy útil para validar formatos (emails, teléfonos, fechas...).

## Patrones básicos

```
.    → cualquier carácter (excepto \n)
\d   → dígito (0-9)
\w   → letra, dígito o guión bajo
\s   → espacio, tabulación o salto de línea
+    → uno o más del patrón anterior
*    → cero o más
?    → cero o uno
{n}  → exactamente n veces
[]   → conjunto de caracteres: [aeiou], [0-9]
^    → inicio de string
$    → fin de string
```

````python
import re

texto = 'Mi email es usuario@ejemplo.com y también uso otro@correo.es'

# re.findall: encuentra todas las coincidencias
emails = re.findall(r'\w+@\w+\.\w+', texto)
print(emails)

# re.search: busca la primera coincidencia
texto2 = 'El código postal es 28001 y el teléfono es 912345678'
m = re.search(r'\d{5}', texto2)
if m:
    print(f"Código postal: {m.group()}")  # 28001
````

````python
# re.sub: sustituir coincidencias
telefono_limpio = re.sub(r'[^\d]', '', '666-123-456')
print(telefono_limpio)   # '666123456'

# re.match: valida el formato completo del string
patron_email = r'^[\w.-]+@[\w.-]+\.\w{2,}$'
print(bool(re.match(patron_email, 'usuario@ejemplo.com')))   # True
print(bool(re.match(patron_email, 'no_es_email')))           # False
````

# 8. `functools` — Herramientas para funciones

Proporciona utilidades de orden superior: crear variantes de funciones, acumuladores y caché automático.

````python
from functools import partial

# partial: crea una función con algunos argumentos ya fijados
def potencia(base, exponente):
    return base ** exponente

cuadrado = partial(potencia, exponente=2)
cubo = partial(potencia, exponente=3)

print(cuadrado(5))   # 25
print(cubo(3))       # 27

# Útil para fijar parámetros de funciones de la biblioteca estándar
print_error = partial(print, '[ERROR]')
print_error('Archivo no encontrado')   # [ERROR] Archivo no encontrado
````

````python
from functools import reduce

# reduce: aplica una función acumulativa a una secuencia
numeros = [1, 2, 3, 4, 5]
producto = reduce(lambda acc, x: acc * x, numeros)
print(producto)   # 120  (1*2*3*4*5)

maximo = reduce(lambda a, b: a if a > b else b, numeros)
print(maximo)   # 5
````

````python
from functools import lru_cache

# lru_cache: caché automático para funciones puras (memoización)
@lru_cache(maxsize=None)
def fibonacci(n: int) -> int:
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(40))          # Rápido gracias al caché
print(fibonacci.cache_info()) # hits, misses y tamaño del caché
````

# 9. `copy` — Copia de objetos

En Python, asignar un objeto mutable a otra variable **no lo copia**, solo crea otra referencia. El módulo `copy` permite hacer copias reales.

````python
import copy

# El problema con la asignación simple
lista_original = [1, 2, [3, 4]]
alias = lista_original            # Solo una referencia, mismo objeto
alias[0] = 99
print(lista_original)             # [99, 2, [3, 4]]  <- también cambió!

# Copia superficial (shallow): copia el contenedor pero no los anidados
lista_original = [1, 2, [3, 4]]
copia_s = copy.copy(lista_original)
copia_s[0] = 99            # No afecta al original
copia_s[2].append(5)       # SÍ afecta: la lista anidada es compartida
print(lista_original)      # [1, 2, [3, 4, 5]]  <- la anidada cambió
print(copia_s)             # [99, 2, [3, 4, 5]]
````

````python
# Copia profunda (deep): copia todo, incluyendo objetos anidados
lista_original = [1, 2, [3, 4]]
copia_d = copy.deepcopy(lista_original)
copia_d[0] = 99
copia_d[2].append(5)
print(lista_original)   # [1, 2, [3, 4]]      <- sin cambios
print(copia_d)          # [99, 2, [3, 4, 5]]  <- independiente

# Regla práctica:
# - Objetos planos (sin anidamiento mutable): copy() es suficiente
# - Objetos con listas/dicts/objetos anidados: usa deepcopy()
````

# 10. `pprint` — Pretty Print

Cuando `print()` muestra estructuras grandes en una sola línea ilegible, `pprint` las formatea de forma clara y jerárquica.

````python
from pprint import pprint, pformat

datos = {
    'usuarios': [
        {'nombre': 'Ana', 'edad': 25, 'hobbies': ['lectura', 'senderismo']},
        {'nombre': 'Luis', 'edad': 30, 'hobbies': ['ajedrez', 'programacion']},
        {'nombre': 'Maria', 'edad': 22, 'hobbies': ['musica', 'cocina', 'viajes']},
    ],
    'total': 3,
    'activo': True
}

print('--- print normal ---')
print(datos)
print()
print('--- pprint ---')
pprint(datos)
````

````python
# Controlar ancho y profundidad
pprint(datos, width=40)    # Fuerza salto de línea (ancho máximo 40 chars)
pprint(datos, depth=1)     # Solo muestra el primer nivel

# pformat: devuelve el string formateado (útil para logs)
texto_formateado = pformat(datos, width=60)
print(type(texto_formateado))   # str
````
